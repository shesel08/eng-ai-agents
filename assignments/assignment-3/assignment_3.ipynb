### Task 1: Drone object detection

```python
from ultralytics import YOLO
```

```python
# Load a pretrained segmentation model like YOLO26n-seg
model = YOLO("yolov8n.pt")  # load a pretrained model (recommended for training)
```

```python
dataset_yaml = "Drone_Object_Detection/data.yaml"
```

```python
results = model.train(data=dataset_yaml, epochs=100, imgsz=640)
```

```python
# After training, you can validate the model's performance on the validation set
results = model.val()
```

```python
import numpy as np
import pandas as pd
import cv2
import shutil
from pathlib import Path
from datasets import load_dataset
```

```python
# All files in a folder
files = list(Path("frames").iterdir())
frames = [cv2.imread(str(p)) for p in Path("frames").glob("*.jpg")]

output_dir = "detections"
output_dir = Path(output_dir)
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
def save_detection_frames(
    frames_dir: str | Path,
    model,
    output_dir: str | Path = "detections",
    conf: float = 0.25,
) -> None:
    """Read all images from frames_dir, run detection, and save frames with at least one detection."""
    frames_dir = Path(frames_dir)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    image_files = sorted(frames_dir.glob("*.jpg")) + sorted(frames_dir.glob("*.png"))
    if not image_files:
        print(f"No image files found in {frames_dir.resolve()}")
        return

    total = len(image_files)
    saved = 0

    for img_path in image_files:
        frame = cv2.imread(str(img_path))
        if frame is None:
            print(f"Could not read {img_path.name}, skipping.")
            continue

        results = model.predict(frame, conf=conf, verbose=False)
        if len(results[0].boxes) == 0:
            continue

        shutil.copy(img_path, output_dir / img_path.name)
        saved += 1

    print(f"Done. {saved}/{total} frames had detections and were saved to '{output_dir.resolve()}'")
```

```python
save_detection_frames("frames", model, output_dir="detections")
```
