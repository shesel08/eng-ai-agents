```python
import numpy as np
import pandas as pd
import cv2
import shutil
from pathlib import Path
from datasets import load_dataset
```

```python
def save_detection_frames(
    frames_dir: str | Path,
    model,
    output_dir: str | Path = "detections",
    conf: float = 0.25,
) -> None:
    """Read all images from frames_dir, run detection, and save frames with at least one detection."""
    frames_dir = Path(frames_dir)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    image_files = sorted(frames_dir.glob("*.jpg"))
    if not image_files:
        print(f"No image files found in {frames_dir.resolve()}")
        return

    total = len(image_files)
    saved = 0

    for img_path in image_files:
        frame = cv2.imread(str(img_path))
        if frame is None:
            print(f"Could not read {img_path.name}, skipping.")
            continue

        results = model.predict(frame, conf=conf, verbose=False)
        if len(results[0].boxes) == 0:
            continue

        shutil.copy(img_path, output_dir / img_path.name)
        saved += 1

    print(f"Done. {saved}/{total} frames had detections and were saved to '{output_dir.resolve()}'")
```

```python
import re
import subprocess
import cv2
import numpy as np
from pathlib import Path
from filterpy.kalman import KalmanFilter
from ultralytics import YOLO
```

```python
# ── Kalman filter factory ────────────────────────────────────────────────────
# State : [cx, cy, vx, vy]   Measurement : [cx, cy]

def make_kalman(cx: float, cy: float, dt: float = 1.0) -> KalmanFilter:
    kf = KalmanFilter(dim_x=4, dim_z=2)
    kf.F = np.array([[1, 0, dt, 0],
                     [0, 1, 0, dt],
                     [0, 0, 1,  0],
                     [0, 0, 0,  1]], dtype=float)
    kf.H = np.array([[1, 0, 0, 0],
                     [0, 1, 0, 0]], dtype=float)
    kf.R *= 10.0    # measurement noise
    kf.Q *= 0.1     # process noise
    kf.P *= 100.0   # initial uncertainty
    kf.x = np.array([[cx], [cy], [0.0], [0.0]], dtype=float)
    return kf
```

```python
# ── IoU helper ───────────────────────────────────────────────────────────────

def iou(b1: list, b2: list) -> float:
    ix1, iy1 = max(b1[0], b2[0]), max(b1[1], b2[1])
    ix2, iy2 = min(b1[2], b2[2]), min(b1[3], b2[3])
    inter = max(0.0, ix2 - ix1) * max(0.0, iy2 - iy1)
    if inter == 0:
        return 0.0
    a1 = (b1[2] - b1[0]) * (b1[3] - b1[1])
    a2 = (b2[2] - b2[0]) * (b2[3] - b2[1])
    return inter / (a1 + a2 - inter)
```

```python
# ── Track ────────────────────────────────────────────────────────────────────

class DroneTrack:
    _next_id: int = 0

    def __init__(self, cx: float, cy: float):
        self.id = DroneTrack._next_id
        DroneTrack._next_id += 1
        self.kf = make_kalman(cx, cy)
        self.misses = 0

    def predict(self) -> None:
        self.kf.predict()

    def update(self, cx: float, cy: float) -> None:
        self.kf.update(np.array([[cx], [cy]], dtype=float))
        self.misses = 0

    def mark_missed(self) -> None:
        self.misses += 1

    @property
    def center(self) -> tuple[float, float]:
        return float(self.kf.x[0]), float(self.kf.x[1])
```

```python
# ── Tracker ──────────────────────────────────────────────────────────────────

def run_tracker(
    frames_dir: str | Path,
    model,
    conf: float = 0.25,
    max_misses: int = 5,
    iou_threshold: float = 0.3,
) -> list[dict]:
    """
    Process every image in frames_dir with YOLO + Kalman filter.

    Returns a list of per-frame dicts:
        img_path : Path
        tracks   : list of {id, cx, cy, bbox (or None if missed)}
    """
    frames_dir = Path(frames_dir)
    image_files = sorted(
        list(frames_dir.glob("*.jpg")) + list(frames_dir.glob("*.png"))
    )
    if not image_files:
        print(f"No images found in {frames_dir.resolve()}")
        return []

    DroneTrack._next_id = 0
    active: list[DroneTrack] = []
    records: list[dict] = []

    for img_path in image_files:
        frame = cv2.imread(str(img_path))
        if frame is None:
            continue

        # 1 – predict
        for t in active:
            t.predict()

        # 2 – detect
        yolo_results = model.predict(frame, conf=conf, verbose=False)
        dets: list[list[float]] = [
            b.xyxy[0].cpu().numpy().tolist()
            for b in yolo_results[0].boxes
        ]

        # 3 – greedy IoU matching
        match: dict[int, int] = {}   # track_idx → det_idx
        if active and dets:
            score = np.zeros((len(active), len(dets)))
            for ti, t in enumerate(active):
                px, py = t.center
                for di, det in enumerate(dets):
                    w, h = det[2] - det[0], det[3] - det[1]
                    pb = [px - w/2, py - h/2, px + w/2, py + h/2]
                    score[ti, di] = iou(pb, det)
            tmp = score.copy()
            while tmp.size > 0:
                ti, di = np.unravel_index(np.argmax(tmp), tmp.shape)
                if tmp[ti, di] < iou_threshold:
                    break
                match[int(ti)] = int(di)
                tmp[ti, :] = -1
                tmp[:, di] = -1

        # 4 – update matched / miss unmatched tracks
        for ti, t in enumerate(active):
            if ti in match:
                det = dets[match[ti]]
                t.update((det[0] + det[2]) / 2, (det[1] + det[3]) / 2)
            else:
                t.mark_missed()

        # 5 – spawn new tracks for unmatched detections
        matched_dets = set(match.values())
        for di, det in enumerate(dets):
            if di not in matched_dets:
                cx, cy = (det[0] + det[2]) / 2, (det[1] + det[3]) / 2
                active.append(DroneTrack(cx, cy))

        # 6 – record frame state
        frame_tracks = []
        for ti, t in enumerate(active):
            frame_tracks.append({
                "id":   t.id,
                "cx":   t.center[0],
                "cy":   t.center[1],
                "bbox": dets[match[ti]] if ti in match else None,
            })
        records.append({"img_path": img_path, "tracks": frame_tracks})

        # 7 – prune dead tracks
        active = [t for t in active if t.misses <= max_misses]

    print(f"Tracked {len(image_files)} frames — {DroneTrack._next_id} total track(s) spawned.")
    return records
```

```python
# ── Video renderer ───────────────────────────────────────────────────────────

# Clip index → source video name
_CLIP_TO_VIDEO = {**{i: "drone_video_1" for i in range(1, 4)},
                  **{i: "drone_video_2" for i in range(4, 25)}}

def _video_name(p: Path) -> str:
    m = re.match(r"frame_(\d+)_", p.name)
    return _CLIP_TO_VIDEO.get(int(m.group(1)), "unknown") if m else "unknown"


def render_tracking_videos(
    records: list[dict],
    output_dir: str | Path = "tracked_output",
    fps: int = 5,
) -> None:
    """
    Produce one MP4 per source video with only frames that have ≥1 detection,
    overlaid with:
      • green  bounding box (detector output)
      • cyan   polyline  (Kalman-estimated trajectory)
      • orange circle    (current Kalman-estimated center)
    """
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    by_video: dict[str, list[dict]] = {}
    for rec in records:
        by_video.setdefault(_video_name(rec["img_path"]), []).append(rec)

    for video_name, recs in by_video.items():
        # Build complete per-track trajectory across all frames
        traj: dict[int, list[tuple[int, int]]] = {}
        for rec in recs:
            for t in rec["tracks"]:
                traj.setdefault(t["id"], []).append((int(t["cx"]), int(t["cy"])))

        running_count: dict[int, int] = {tid: 0 for tid in traj}

        sample = cv2.imread(str(recs[0]["img_path"]))
        h, w = sample.shape[:2]

        tmp_path = output_dir / f"{video_name}_tracked_tmp.mp4"
        writer = cv2.VideoWriter(
            str(tmp_path),
            cv2.VideoWriter_fourcc(*"mp4v"),
            fps,
            (w, h),
        )

        for rec in recs:
            for t in rec["tracks"]:
                running_count[t["id"]] = running_count.get(t["id"], 0) + 1

            if not any(t["bbox"] is not None for t in rec["tracks"]):
                continue   # skip frames with no fresh detection

            frame = cv2.imread(str(rec["img_path"]))
            if frame is None:
                continue

            for t in rec["tracks"]:
                tid  = t["id"]
                cx, cy = int(t["cx"]), int(t["cy"])
                pts = traj[tid][:running_count.get(tid, len(traj[tid]))]

                # Trajectory polyline
                if len(pts) > 1:
                    poly = np.array(pts, dtype=np.int32).reshape(-1, 1, 2)
                    cv2.polylines(frame, [poly], isClosed=False,
                                  color=(0, 255, 255), thickness=2)

                # Detector bounding box
                if t["bbox"] is not None:
                    x1, y1, x2, y2 = (int(v) for v in t["bbox"])
                    cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                    cv2.putText(frame, f"T{tid}", (x1, max(y1 - 6, 12)),
                                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

                # Kalman-predicted center
                cv2.circle(frame, (cx, cy), 5, (0, 128, 255), -1)

            writer.write(frame)

        writer.release()

        # Re-encode to H.264 with ffmpeg
        final_path = output_dir / f"{video_name}_tracked.mp4"
        try:
            subprocess.run(
                ["ffmpeg", "-y", "-i", str(tmp_path),
                 "-vcodec", "libx264", "-crf", "23", "-pix_fmt", "yuv420p",
                 str(final_path)],
                check=True, capture_output=True,
            )
            tmp_path.unlink()
            print(f"Saved: {final_path}")
        except subprocess.CalledProcessError as e:
            print(f"ffmpeg re-encode failed — keeping raw output at {tmp_path}")
            print(e.stderr.decode())
```

```python
# ── Run ──────────────────────────────────────────────────────────────────────

records = run_tracker("detections", model)
render_tracking_videos(records, output_dir="tracked_output")
```
